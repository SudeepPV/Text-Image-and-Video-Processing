## Text Data in .JSON

```python
import json

# ---------------------------------------
# 1. Create JSON data (Python dictionary)
# ---------------------------------------
student = {
    "id": 101,
    "name": "Alice",
    "age": 22,
    "department": "Computer Science",
    "subjects": [
        "Machine Learning",
        "Image Processing",
        "Deep Learning"
    ],
    "cgpa": 8.92,
    "hosteller": True
}

# ---------------------------------------
# 2. Save (Write) JSON data to a file
# ---------------------------------------
filename = "student.json"

with open(filename, "w") as file:
    json.dump(student, file, indent=4)

print("JSON file created successfully.")
```

```python
# ---------------------------------------
# 3. Load (Read) JSON data from file
# ---------------------------------------
with open(filename, "r") as file:
    data = json.load(file)

print("\nLoaded JSON Data:")
print(data)
```

```python
# ---------------------------------------
# 4. Access individual values
# ---------------------------------------
print("\nStudent Name :", data["name"])
print("Department   :", data["department"])
print("First Subject:", data["subjects"][0])
```

```python
# ---------------------------------------
# 5. Modify (Update) JSON data
# ---------------------------------------
data["cgpa"] = 9.15
data["subjects"].append("Computer Vision")
data["email"] = "alice@example.com"

# ---------------------------------------
# 6. Save updated data back to the file
# ---------------------------------------
with open(filename, "w") as file:
    json.dump(data, file, indent=4)

print("\nUpdated JSON saved.")

# ---------------------------------------
# 7. Read the updated file
# ---------------------------------------
with open(filename, "r") as file:
    updated_data = json.load(file)

print("\nUpdated JSON Data:")
print(json.dumps(updated_data, indent=4))
```

## Convert a Python object to a JSON string

```python
import json

person = {
    "name": "John",
    "age": 30,
    "city": "New York"
}

json_string = json.dumps(person, indent=4)

print(json_string)
```

## Convert a JSON string to a Python object

```python
import json

json_text = '''
{
    "name": "John",
    "age": 30,
    "city": "New York"
}
'''

person = json.loads(json_text)

print(person)
print(person["name"])
```

## Text Data in .XML

```python
import xml.etree.ElementTree as ET

# =====================================================
# CREATE XML DATA
# =====================================================

# Create the root element
root = ET.Element("students")

# Create first student
student1 = ET.SubElement(root, "student", id="101")

ET.SubElement(student1, "name").text = "Alice"
ET.SubElement(student1, "age").text = "21"
ET.SubElement(student1, "department").text = "Computer Science"
ET.SubElement(student1, "cgpa").text = "8.75"

# Create second student
student2 = ET.SubElement(root, "student", id="102")

ET.SubElement(student2, "name").text = "Bob"
ET.SubElement(student2, "age").text = "22"
ET.SubElement(student2, "department").text = "Electronics"
ET.SubElement(student2, "cgpa").text = "9.10"

# =====================================================
# SAVE XML TO FILE
# =====================================================

tree = ET.ElementTree(root)

tree.write(
    "students.xml",
    encoding="utf-8",
    xml_declaration=True
)

print("XML file created successfully.")
```

```python
# =====================================================
# LOAD XML FILE
# =====================================================

tree = ET.parse("students.xml")
root = tree.getroot()

print("\nLoaded XML successfully.")

# =====================================================
# READ XML DATA
# =====================================================

print("\nStudent Details")
print("-" * 40)

for student in root.findall("student"):

    sid = student.get("id")
    name = student.find("name").text
    age = student.find("age").text
    dept = student.find("department").text
    cgpa = student.find("cgpa").text

    print(f"ID         : {sid}")
    print(f"Name       : {name}")
    print(f"Age        : {age}")
    print(f"Department : {dept}")
    print(f"CGPA       : {cgpa}")
    print()
```

```python
# =====================================================
# UPDATE EXISTING DATA
# =====================================================
for student in root.findall("student"):

    if student.get("id") == "101":
        student.find("cgpa").text = "9.00"

# Save changes
tree.write(
    "students_updated.xml",
    encoding="utf-8",
    xml_declaration=True
)

print("Updated XML saved as students_updated.xml")
```

```python
# =====================================================
# LOAD XML FILE
# =====================================================
tree = ET.parse("students_updated.xml")
root = tree.getroot()

print("\nLoaded XML successfully.")

# =====================================================
# READ XML DATA
# =====================================================
print("\nStudent Details")
print("-" * 40)

for student in root.findall("student"):
    sid = student.get("id")
    name = student.find("name").text
    age = student.find("age").text
    dept = student.find("department").text
    cgpa = student.find("cgpa").text

    print(f"ID         : {sid}")
    print(f"Name       : {name}")
    print(f"Age        : {age}")
    print(f"Department : {dept}")
    print(f"CGPA       : {cgpa}")
    print()
```

## Text data in .csv

```python
import csv

# -------------------------------------
# 1. Create and Write a CSV File
# -------------------------------------

filename = "students.csv"

data = [
    ["ID", "Name", "Age", "Department"],
    [101, "Alice", 20, "CSE"],
    [102, "Bob", 21, "ECE"],
    [103, "Charlie", 22, "EEE"]
]

with open(filename, "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerows(data)

print("CSV file created successfully.")
```

```python
# -------------------------------------
# 2. Read the CSV File
# -------------------------------------

print("\nReading CSV File:")

with open(filename, "r", encoding="utf-8") as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

```python
# -------------------------------------
# 3. Append a New Row
# -------------------------------------

new_record = [104, "David", 23, "ME"]

with open(filename, "a", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(new_record)

print("\nNew record added.")
```

```python
# -------------------------------------
# 4. Load CSV into a List
# -------------------------------------

loaded_data = []

with open(filename, "r", encoding="utf-8") as file:
    reader = csv.reader(file)
    for row in reader:
        loaded_data.append(row)

print("\nLoaded Data:")
print(loaded_data)
```

```python
# -------------------------------------
# 5. Modify the Data
# -------------------------------------

loaded_data[2][2] = "25"      # Change Bob's age
```

```python
# -------------------------------------
# 6. Save the Modified Data
# -------------------------------------

with open(filename, "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerows(loaded_data)

print("\nModified data saved successfully.")
```

## Text data in .csv (DictReader and DictWriter)

```python
import csv

filename = "employees.csv"

# -------------------------------------
# Create and Save
# -------------------------------------

employees = [
    {"ID": 1, "Name": "John", "Salary": 50000},
    {"ID": 2, "Name": "Emma", "Salary": 62000},
    {"ID": 3, "Name": "Ryan", "Salary": 58000}
]

with open(filename, "w", newline="", encoding="utf-8") as file:
    fieldnames = ["ID", "Name", "Salary"]

    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(employees)

print("Employee file saved.")

# -------------------------------------
# Load
# -------------------------------------

with open(filename, "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)

    for row in reader:
        print(row)
```

## Text data in.txt

```python
# ==========================================
# Text File Operations in Python (.txt)
# ==========================================

# File name
filename = "sample.txt"

# ------------------------------------------
# 1. Create and Write to a Text File
# ------------------------------------------
with open(filename, "w") as file:
    file.write("Hello, World!\n")
    file.write("Welcome to Python file handling.\n")
    file.write("This is the third line.\n")

print("File created and data written successfully.")
```

```python
# ------------------------------------------
# 2. Read the Entire File
# ------------------------------------------
with open(filename, "r") as file:
    content = file.read()

print("\nEntire File Content:")
print(content)
```

```python
# ------------------------------------------
# 3. Read File Line by Line
# ------------------------------------------
print("Reading Line by Line:")
with open(filename, "r") as file:
    for line in file:
        print(line.strip())
```

```python
# ------------------------------------------
# 4. Read All Lines into a List
# ------------------------------------------
with open(filename, "r") as file:
    lines = file.readlines()

print("\nList of Lines:")
print(lines)
```

```python
# ------------------------------------------
# 5. Append New Data
# ------------------------------------------
with open(filename, "a") as file:
    file.write("This line was appended later.\n")

print("\nData appended successfully.")
```

```python
# ------------------------------------------
# 6. Load the Updated File
# ------------------------------------------
with open(filename, "r") as file:
    updated_content = file.read()

print("\nUpdated File Content:")
print(updated_content)
```

```python
# ------------------------------------------
# 7. Save Another Copy
# ------------------------------------------
backup_filename = "sample_backup.txt"

with open(backup_filename, "w") as backup:
    backup.write(updated_content)

print(f"\nBackup saved as '{backup_filename}'")
```
